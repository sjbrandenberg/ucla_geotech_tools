```python
def get_zchol_zkrig(x, z, ind, ind_chol, ind_krig, max_dist):
    # Select first point if empty arrays are passed in
    if(len(ind_chol) == 0):
        ind_chol = np.append(ind_chol, 0)

    mask = ~np.isin(ind, np.hstack((ind_chol, ind_krig)))
    dist = np.sqrt((x[ind_chol[-1]] - x) ** 2 + (z[ind_chol[-1]] - z) ** 2)

    # If we have found an isolated set where all points are beyond max_dist, pick a new point that is not already selected
    if(np.min(dist[mask]) >= max_dist):
        ind_chol = np.append(ind_chol, ind[mask][0])
    else:
        ind_chol = np.append(ind_chol, np.argwhere(dist == np.max(dist[mask][dist[mask] < max_dist]))[0])
    ind_krig = np.append(ind_krig, ind[(dist < max_dist) & ~np.isin(ind, np.hstack((ind_chol, ind_krig)))])
    
    return ([ind_chol, ind_krig)

N = 1000
ind = np.arange(N)
ind_chol = np.asarray([], dtype=int)
ind_krig = np.asarray([], dtype=int)
x = np.random.rand(N) * 10
z = np.random.rand(N) * 10

max_dist = 2
while (len(ind_chol) + len(ind_krig) < N):
    ind_chol, ind_krig = get_zchol_zkrig(x, z, ind, ind_chol, ind_krig, max_dist)
```

```python
def get_zchol_zkrig(x, z, max_dist):
    ind = np.arange(len(x))
    ind_chol = np.asarray([0], dtype=int)
    ind_krig = np.asarray([], dtype=int)

    while(len(ind_chol) + len(ind_krig) < len(ind)):
        mask = ~np.isin(ind, np.hstack((ind_chol, ind_krig)))
        dist = np.sqrt((x[ind_chol[-1]] - x) ** 2 + (z[ind_chol[-1]] - z) ** 2)
    
        # If we have found an isolated set where all points are beyond max_dist, pick a new point that is not already selected
        if(np.min(dist[mask]) >= max_dist):
            ind_chol = np.append(ind_chol, ind[mask][0])
        else:
            ind_chol = np.append(ind_chol, np.argwhere(dist == np.max(dist[mask][dist[mask] < max_dist]))[0])
        ind_krig = np.append(ind_krig, ind[(dist < max_dist) & ~np.isin(ind, np.hstack((ind_chol, ind_krig)))])
        
    return (ind_chol, x[ind_chol], z[ind_chol], ind_krig, x[ind_krig], z[ind_krig])
```
